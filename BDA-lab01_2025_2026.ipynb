### 7. Dictionaries:

### Imputation of Missing Values in Pandas

Missing values, often represented as `NaN` (Not a Number) in pandas, can negatively impact data analysis and machine learning models. Imputation is the process of replacing these missing values with substituted values.

First, let's confirm the missing values in our `df` DataFrame:

```
df = purchases
print("Missing values before imputation:")
display(df.isnull().sum())
```

From the output, we can see that the `name` and `price` columns have missing values. Here are some common imputation strategies:

1.  **Mean Imputation**: Replace missing values with the mean of the column. This is suitable for numerical data but can distort the distribution and relationships.
2.  **Median Imputation**: Replace missing values with the median of the column. This is more robust to outliers than mean imputation.
3.  **Mode Imputation**: Replace missing values with the most frequent value (mode) of the column. This is often used for categorical data or discrete numerical data.
4.  **Forward Fill (`ffill`)**: Propagate the last valid observation forward to next missing ones. Useful for time series data.
5.  **Backward Fill (`bfill`)**: Propagate the next valid observation backward to previous missing ones. Also useful for time series data.
6.  **Constant Value Imputation**: Replace missing values with a specific constant (e.g., 0, 'Unknown').
7.  **Dropping Rows/Columns**: If the number of missing values is small or the row/column is not critical, you might choose to drop them. (Note: Dropping rows with missing essential identifiers like 'name' was already demonstrated in a previous cell: `LHfovIlNVf5G`).

Let's demonstrate some of these on our `df`:

```
# Original DataFrame for comparison
print("Original DataFrame with missing values:")
display(df)

# Median Imputation for 'price'
# This was already demonstrated in cell -aZXP3zwU5Cu. Let's re-run and explain it.
median_price = df["price"].median(skipna=True)
imputed_median_price_df = df.copy()
imputed_median_price_df["price"] = imputed_median_price_df["price"].fillna(median_price)
print("\nDataFrame after Median Imputation for 'price' (using value: ", median_price, "):")
display(imputed_median_price_df)

# Mean Imputation for 'price'
mean_price = df["price"].mean(skipna=True)
imputed_mean_price_df = df.copy()
imputed_mean_price_df["price"] = imputed_mean_price_df["price"].fillna(mean_price)
print("\nDataFrame after Mean Imputation for 'price' (using value: ", round(mean_price, 2), "):")
display(imputed_mean_price_df)

# Mode Imputation for 'name' (categorical data)
# Note: mode can return multiple values if there's a tie, so we take the first one.
mode_name = df["name"].mode()[0]
imputed_mode_name_df = df.copy()
imputed_mode_name_df["name"] = imputed_mode_name_df["name"].fillna(mode_name)
print("\nDataFrame after Mode Imputation for 'name' (using value: '", mode_name, "'):")
display(imputed_mode_name_df)

# Forward Fill (ffill) for 'price'
ffill_df = df.copy()
ffill_df["price"] = ffill_df["price"].ffill()
print("\nDataFrame after Forward Fill for 'price':")
display(ffill_df)

# Backward Fill (bfill) for 'price'
bfill_df = df.copy()
bfill_df["price"] = bfill_df["price"].bfill()
print("\nDataFrame after Backward Fill for 'price':")
display(bfill_df)
```

The choice of imputation strategy depends on the nature of your data and the specific problem you are trying to solve. For the `price` column, median or mean imputation are common choices, while for `name`, filling with a mode or a placeholder like 'Unknown' (as shown in cell `y5c3qeCNUfVf`) is more appropriate.

## DataFrame slicing, selecting, extracting

* Up until now we've focused on some basic summaries of our data. We've learned about simple column extraction using single brackets, and we imputed null values in a column using fillna(). Below are the other methods of slicing, selecting, and extracting you'll need to use constantly.

* It's important to note that, although many methods are the same, DataFrames and Series have different attributes, so you'll need be sure to know which type you are working with or else you will receive attribute errors.

* Let's look at working with columns first.

#### By column

* You already saw how to extract a column using square brackets like this:

```
genre_col = movies_df['Genre']

type(genre_col)
```

* This will return a **Series**. To extract a column as a DataFrame, you need to pass a list of column names. In our case that's just a single column:

```
genre_col = movies_df[['Genre']]

type(genre_col)
```

* Since it's just a list, adding another column name is easy:

```
subset = movies_df[['Genre', 'Rating']]

subset.head()
```

#### Conditional selections

* We’ve gone over how to select columns and rows, but what if we want to make a conditional selection?


* For example, what if we want to filter our movies DataFrame to show only films directed by Ridley Scott or films with a rating greater than or equal to 8.0?


* To do that, we take a column from the DataFrame and apply a Boolean condition to it. Here's an example of a Boolean condition:

```
condition = (movies_df['Director'] == "Ridley Scott")

condition.head()
```

* We want to filter out all movies not directed by Ridley Scott, in other words, we don’t want the False films. To return the rows where that condition is True we have to pass this operation into the DataFrame:

```
movies_df['Director'].unique()
```

```
# movies_df[condition]
# OR
movies_df[(movies_df['Director'] == "Ridley Scott") & (movies_df["Rating"] > 7.5)]
# movies_df[movies_df.Director == 'Ridley Scott']
```

* You can get used to looking at these conditionals by reading it like: **Select movies_df where movies_df director equals Ridley Scott.**


* Let's look at conditional selections using numerical values by filtering the DataFrame by ratings:

```
movies_df[movies_df.Rating >= 8.6].head(3)
```

* We can make some richer conditionals by using logical operators **| for "or" and & for "and"**.

* Let's filter the the DataFrame to show only movies by Christopher Nolan OR Ridley Scott:

```
movies_df[(movies_df['Director'] == 'Christopher Nolan') | (movies_df['Director'] == 'Ridley Scott')].head(15)
```

```
movies_df[(movies_df['Director'] == 'Christopher Nolan') & (movies_df['Rating'] >= 6)].head(15)
```

# grouping, pivot and crosstab

```
# Grouping
import pandas as pd
df = pd.DataFrame({
    "date": pd.to_datetime([
        "2025-01-01","2025-01-01","2025-01-02","2025-01-02","2025-01-02","2025-01-03"
    ]),
    "region": ["N","S","N","S","S","N"],
    "product": ["A","A","B","A","B","A"],
    "units": [10, 5, 8, 7, 3, 4],
    "price": [2.0, 2.5, 3.0, 2.0, 3.5, 2.0],
    "customer": ["c1","c2","c1","c3","c2","c4"]
})
df["revenue"] = df["units"] * df["price"]
df

df.groupby("region", as_index=False)["revenue"].sum().sort_values("revenue", ascending=False)
```

```
df.head()
```

```
# totals by region & product with multiple metrics
summary = (
    df.groupby(["region","product"], as_index=False)
      .agg(
          total_units = ("units", "sum"),
          avg_price   = ("price", "mean"),
          orders      = ("product", "count"),
          customers   = ("customer", "nunique"),
          revenue     = ("revenue", "sum")
      )
      .sort_values(["region","revenue"], ascending=[True, False])
)
summary
```

```
def wavg(x, w):
    return (x * w).sum() / w.sum()

wavg_price = (
    df.groupby(["region","product"])
      .apply(lambda g: wavg(g["price"], g["units"]))
      .reset_index(name="wavg_price")
)
wavg_price
```

```
# size: number of rows per group (ignores NA issues)
df.groupby("region").size().reset_index(name="rows")

# count: non-null count per column
df.groupby("region")[["units","price"]].count().reset_index()
```

```
# top product (by revenue) per region
top_prod = (
    df.groupby(["region","product"], as_index=False)["revenue"].sum()
      .sort_values(["region","revenue"], ascending=[True, False])
      .groupby("region").head(1)
)
top_prod
```

```
# revenue matrix: rows=region, columns=product
pd.pivot_table(
    df, values="revenue", index="region", columns="product", aggfunc="sum", fill_value=0
)
```

```
# revenue per day
daily = df.groupby("date", as_index=False)["revenue"].sum()

# revenue per month (safe for irregular dates)
monthly = (
    df.assign(month=df["date"].dt.to_period("M").dt.to_timestamp())
      .groupby("month", as_index=False)["revenue"].sum()
)
daily, monthly
```

```
# Crosstab: count of products per region
pd.crosstab(df["region"], df["product"])
```
